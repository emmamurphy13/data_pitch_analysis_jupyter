```python
import pandas as pd
import re

# Input files
rural_file = "rural_nys_school.csv"   # or "rural_nys_schools_csv.csv" if that is your actual filename
schools_file = "NYS_SCHOOLS.csv"

# Output file
out_file = "NYS_SCHOOLS_rural_counties.csv"

def norm_county(x):
    if pd.isna(x):
        return ""
    s = str(x).upper().strip()
    s = s.replace("&", "AND")
    s = re.sub(r"\bST\b\.?", "SAINT", s)
    s = re.sub(r"\bCOUNTY\b", "", s)
    s = re.sub(r"[^A-Z0-9 ]+", " ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

# Load
rural = pd.read_csv(rural_file, dtype=str)
schools = pd.read_csv(schools_file, dtype=str)

# Detect county columns
rural_county_col = "County_Name" if "County_Name" in rural.columns else "COUNTY"
schools_county_col = "County_Name" if "County_Name" in schools.columns else "COUNTY"

if rural_county_col not in rural.columns:
    raise ValueError(f"County column not found in rural file. Columns: {list(rural.columns)}")
if schools_county_col not in schools.columns:
    raise ValueError(f"County column not found in schools file. Columns: {list(schools.columns)}")

# County set from rural file
rural_counties = set(rural[rural_county_col].map(norm_county).dropna())

# Filter NYS_SCHOOLS to only those counties
schools["_county_norm"] = schools[schools_county_col].map(norm_county)
filtered = schools[schools["_county_norm"].isin(rural_counties)].copy()
filtered = filtered.drop(columns=["_county_norm"])

df
```

```python
#### THIS CODE COMBINES ADDS A COUNTY COLLUMN TO THE GRADUATION RATES CSV

import pandas as pd
import re
from IPython.display import display

# Paths
grad_path = "/Users/emmamurphy/dataProjects/migranted/Data/2023-24_ELL_Graduation_Rate.csv"
county_ref_path = "/Users/emmamurphy/dataProjects/migranted/Data/ell_school_ID_with_county.csv"
out_path = "/Users/emmamurphy/dataProjects/migranted/Data/2023-24_ELL_Graduation_Rate_with_county.csv"

# Load
grad = pd.read_csv(grad_path, dtype=str)
ref = pd.read_csv(county_ref_path, dtype=str)

# Clean column names
grad.columns = grad.columns.str.strip()
ref.columns = ref.columns.str.strip()

# Ensure key columns are strings
for c in ["INSTITUTION_ID", "ENTITY_CD", "ENTITY_NAME"]:
    if c in grad.columns:
        grad[c] = grad[c].fillna("").astype(str).str.strip()
    if c in ref.columns:
        ref[c] = ref[c].fillna("").astype(str).str.strip()

# Ensure COUNTY exists in output
if "COUNTY" not in grad.columns:
    grad["COUNTY"] = ""

# ---------- 1) Match by INSTITUTION_ID ----------
id_map = (
    ref.loc[ref["INSTITUTION_ID"] != "", ["INSTITUTION_ID", "COUNTY"]]
    .dropna(subset=["COUNTY"])
    .drop_duplicates(subset=["INSTITUTION_ID"])
    .set_index("INSTITUTION_ID")["COUNTY"]
)
missing = grad["COUNTY"].eq("")
grad.loc[missing, "COUNTY"] = grad.loc[missing, "INSTITUTION_ID"].map(id_map).fillna("")

# ---------- 2) Match by ENTITY_CD ----------
cd_map = (
    ref.loc[ref["ENTITY_CD"] != "", ["ENTITY_CD", "COUNTY"]]
    .dropna(subset=["COUNTY"])
    .drop_duplicates(subset=["ENTITY_CD"])
    .set_index("ENTITY_CD")["COUNTY"]
)
missing = grad["COUNTY"].eq("")
grad.loc[missing, "COUNTY"] = grad.loc[missing, "ENTITY_CD"].map(cd_map).fillna("")

# ---------- 3) Fallback: match by normalized ENTITY_NAME ----------
def norm_name(x):
    s = "" if pd.isna(x) else str(x).upper().strip()
    s = re.sub(r"[^A-Z0-9 ]+", " ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

grad["__name_key"] = grad["ENTITY_NAME"].map(norm_name)
ref["__name_key"] = ref["ENTITY_NAME"].map(norm_name)

name_map = (
    ref.loc[ref["__name_key"] != "", ["__name_key", "COUNTY"]]
    .dropna(subset=["COUNTY"])
    .drop_duplicates(subset=["__name_key"])
    .set_index("__name_key")["COUNTY"]
)
missing = grad["COUNTY"].eq("")
grad.loc[missing, "COUNTY"] = grad.loc[missing, "__name_key"].map(name_map).fillna("")

# Cleanup temp column
grad.drop(columns=["__name_key"], inplace=True)

# Save
grad.to_csv(out_path, index=False)

# Quick check
print("Saved:", out_path)
print("Total rows:", len(grad))
print("Rows with COUNTY:", (grad["COUNTY"].str.strip() != "").sum())
print("Rows missing COUNTY:", (grad["COUNTY"].str.strip() == "").sum())

# Small excerpt
display(grad[["ENTITY_NAME", "INSTITUTION_ID", "ENTITY_CD", "COUNTY"]].head(15))
```

```python
import pandas as pd
import re
from IPython.display import display

# Paths
grad_path = "/Users/emmamurphy/dataProjects/migranted/Data/2023-24_ELL_Graduation_Rate_with_county.csv"
rural_path = "/Users/emmamurphy/dataProjects/migranted/Data/rural_counties.csv"
out_path = "/Users/emmamurphy/dataProjects/migranted/Data/2023-24_ELL_Graduation_Rate_rural_only.csv"

def norm_county(x):
    if pd.isna(x):
        return ""
    s = str(x).upper().strip()
    s = s.replace(".", "")          # ST. -> ST
    s = re.sub(r"\bCOUNTY\b", "", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

# Load files
grad = pd.read_csv(grad_path, dtype=str)
rural = pd.read_csv(rural_path, dtype=str)

# Clean columns
grad.columns = grad.columns.str.strip()
rural.columns = rural.columns.str.strip()

if "COUNTY" not in grad.columns:
    raise ValueError("COUNTY column not found in graduation file.")
if "County_Name" not in rural.columns:
    raise ValueError("County_Name column not found in rural_counties.csv.")

# Build allowed rural county set
allowed = set(rural["County_Name"].map(norm_county).dropna())

# Filter graduation rows to only allowed counties
grad["_county_norm"] = grad["COUNTY"].map(norm_county)
rural_only = grad[grad["_county_norm"].isin(allowed)].copy()
rural_only.drop(columns=["_county_norm"], inplace=True)

# Save
rural_only.to_csv(out_path, index=False)

# Checks
print("Allowed rural counties:", len(allowed))
print("Input rows:", len(grad))
print("Output rows:", len(rural_only))
print("Albany rows kept:", (rural_only["COUNTY"].astype(str).str.upper().str.strip() == "ALBANY").sum())
print("Saved:", out_path)

# Small excerpt
display(rural_only[["ENTITY_NAME", "INSTITUTION_ID", "ENTITY_CD", "COUNTY"]].head(15))
```

```python
# CHECKING IF ALL 33 COUNTIES ARE INCLUDED

# Distinct counties in filtered output
included_norm = sorted(set(rural_only["COUNTY"].map(norm_county)) - {""})
included_raw = sorted(set(rural_only["COUNTY"].astype(str).str.strip()) - {""})

print("Distinct counties included (normalized):", len(included_norm))
print("Distinct counties included (raw):", len(included_raw))

# Compare to allowed rural county list
missing_from_output = sorted(set(allowed) - set(included_norm))
unexpected_in_output = sorted(set(included_norm) - set(allowed))

print("Allowed rural counties total:", len(allowed))
print("Allowed counties missing from output:", len(missing_from_output))
print("Unexpected counties in output:", len(unexpected_in_output))

print("\nIncluded counties (normalized):")
print(included_norm)

if missing_from_output:
    print("\nMissing allowed counties:")
    print(missing_from_output)

if unexpected_in_output:
    print("\nUnexpected counties:")
    print(unexpected_in_output)

# Optional: rows per county
county_counts = (
    rural_only.assign(COUNTY_NORM=rural_only["COUNTY"].map(norm_county))
             .groupby("COUNTY_NORM", dropna=False)
             .size()
             .reset_index(name="ROW_COUNT")
             .sort_values("ROW_COUNT", ascending=False)
)
display(county_counts.head(50))
```

```python
# FINDING OUT WHICH TWO AREN'T INCLUDED AND WHY
missing = sorted(set(allowed) - set(included_norm))
print("Missing counties:", missing)

# Check if those counties exist anywhere in grad file before filtering
grad_check = grad.copy()
grad_check["COUNTY_NORM"] = grad_check["COUNTY"].map(norm_county)

for c in missing:
    rows_with_county = grad_check[grad_check["COUNTY_NORM"] == c]
    print(f"\n{c}: rows in grad with this COUNTY =", len(rows_with_county))
    if len(rows_with_county) == 0:
        # find possible rows by ENTITY_NAME hint
        hint = c.replace("ST ", "ST. ")
        name_hits = grad_check[
            grad_check["ENTITY_NAME"].astype(str).str.upper().str.contains(c, na=False) |
            grad_check["ENTITY_NAME"].astype(str).str
                .upper().str.contains(hint, na=False)
        ]
        print("Possible ENTITY_NAME hits with missing/other county:", len(name_hits))
        display(name_hits[["ENTITY_NAME","INSTITUTION_ID","ENTITY_CD","COUNTY"]].head(20))
```

```python
#### ADDING IN ST. LAWRENCE COUNTY

import pandas as pd
import re
from IPython.display import display

grad_path = "/Users/emmamurphy/dataProjects/migranted/Data/2023-24_ELL_Graduation_Rate_with_county.csv"
rural_path = "/Users/emmamurphy/dataProjects/migranted/Data/rural_counties.csv"
out_path = "/Users/emmamurphy/dataProjects/migranted/Data/2023-24_ELL_Graduation_Rate_rural_only.csv"

def norm_county(x):
    if pd.isna(x):
        return ""
    s = str(x).upper().strip()
    s = s.replace(".", "")
    s = re.sub(r"\bCOUNTY\b", "", s)
    s = re.sub(r"\s+", " ", s).strip()
    # treat ST and SAINT the same
    s = s.replace("ST LAWRENCE", "SAINT LAWRENCE")
    return s

# Load files
grad = pd.read_csv(grad_path, dtype=str)
rural = pd.read_csv(rural_path, dtype=str)

grad.columns = grad.columns.str.strip()
rural.columns = rural.columns.str.strip()

if "COUNTY" not in grad.columns:
    raise ValueError("COUNTY column not found in graduation file.")
if "County_Name" not in rural.columns:
    raise ValueError("County_Name column not found in rural_counties.csv.")

# Build rural county keep-list
allowed = set(rural["County_Name"].map(norm_county).dropna())

# Ensure St. Lawrence is included in the keep-list
allowed.add("SAINT LAWRENCE")

# Filter graduation file
grad["_county_norm"] = grad["COUNTY"].map(norm_county)
rural_only = grad[grad["_county_norm"].isin(allowed)].copy()
rural_only.drop(columns=["_county_norm"], inplace=True)

# Save
rural_only.to_csv(out_path, index=False)

# Checks
print("Allowed counties:", len(allowed))
print("Input rows:", len(grad))
print("Output rows:", len(rural_only))
print("St. Lawrence rows kept:", rural_only["COUNTY"].astype(str).str.upper().str.contains("ST|SAINT", na=False).sum())
print("Saved:", out_path)

display(rural_only[["ENTITY_NAME", "INSTITUTION_ID", "ENTITY_CD", "COUNTY"]].head(15))
```

```python
#### NON-RURAL ENROLLMENT

import pandas as pd

# Paths
full_path = "/Users/emmamurphy/dataProjects/migranted/2023-24_ELL_Enrollment.csv"
filtered_path = "/Users/emmamurphy/dataProjects/migranted/Data/Cleaned data and Analysis/enrollment.csv"
out_path = "/Users/emmamurphy/dataProjects/migranted/Data/Cleaned data and Analysis/enrollment_not_in_enrollment.csv"

# Load
full = pd.read_csv(full_path, dtype=str)
filtered = pd.read_csv(filtered_path, dtype=str)

# Clean column names
full.columns = full.columns.str.strip()
filtered.columns = filtered.columns.str.strip()

# Pick best school key for exclusion
key_candidates = ["INSTITUTION_ID", "ENTITY_CD", "ENTITY_NAME"]
key = next((k for k in key_candidates if k in full.columns and k in filtered.columns), None)
if key is None:
    raise ValueError(f"No shared key found. Need one of: {key_candidates}")

full[key] = full[key].fillna("").astype(str).str.strip()
filtered[key] = filtered[key].fillna("").astype(str).str.strip()

# Exclude schools already in enrollment.csv
exclude_keys = set(filtered[key])
remaining = full[~full[key].isin(exclude_keys)].copy()

# Apply same column filter/order as enrollment.csv
cols_to_keep = [c for c in filtered.columns if c in remaining.columns]
remaining = remaining[cols_to_keep].copy()

# If enrollment.csv is non-zero ELL only, apply same rule
if "NUM_TOTAL_ELL" in remaining.columns and "NUM_TOTAL_ELL" in filtered.columns:
    f_num = pd.to_numeric(filtered["NUM_TOTAL_ELL"], errors="coerce")
    if (f_num.fillna(0) > 0).all():
        remaining["NUM_TOTAL_ELL"] = pd.to_numeric(remaining["NUM_TOTAL_ELL"], errors="coerce").fillna(0)
        remaining = remaining[remaining["NUM_TOTAL_ELL"] > 0].copy()

# If enrollment.csv is sorted by NUM_TOTAL_ELL desc, keep same behavior
if "NUM_TOTAL_ELL" in remaining.columns:
    remaining["NUM_TOTAL_ELL"] = pd.to_numeric(remaining["NUM_TOTAL_ELL"], errors="coerce").fillna(0)
    remaining = remaining.sort_values("NUM_TOTAL_ELL", ascending=False)

# Save
remaining.to_csv(out_path, index=False)

print("Key used:", key)
print("Rows in full:", len(full))
print("Rows excluded (already in enrollment.csv):", len(full) - len(remaining))
print("Rows saved:", len(remaining))
print("Saved to:", out_path)
remaining.head(20)
```
